```python
import pandas as pd
import numpy as np
import os

data_path = "../data/raw/"

churn = pd.read_csv(data_path + "Churn_Modelling.csv")
findex = pd.read_csv(data_path + "GlobalFindexDatabase2025.csv")
mobile = pd.read_csv(data_path + "active-mobile-money-accounts.csv")

print("All three datasets loaded successfully!")
```

```python
datasets = {
    "Churn Modelling": churn,
    "Global Findex": findex,
    "Active Mobile Money": mobile
}

for name, df in datasets.items():
    print("\n", "=" * 50)
    print(name)
    print("=" * 50)

    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())

    print("\nMissing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Data types:")
print(churn.dtypes)

print("\nUnique values in categorical columns:")
print("Geography:", churn["Geography"].unique())
print("Gender:", churn["Gender"].unique())

print("\nTarget variable distribution:")
print(churn["Exited"].value_counts())

print("\nNumerical summary:")
display(churn.describe())
```

```python
churn_clean = churn.drop(
    columns=["RowNumber", "CustomerId", "Surname"]
)

print("Original shape:", churn.shape)
print("Cleaned shape:", churn_clean.shape)

display(churn_clean.head())
```

```python
print("Missing values:")
print(churn_clean.isnull().sum())

print("\nDuplicate rows:", churn_clean.duplicated().sum())

print("\nData types:")
print(churn_clean.dtypes)
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

churn_clean.to_csv(
    "../data/processed/churn_cleaned.csv",
    index=False
)

print("Churn dataset cleaned and saved successfully!")
```

```python
print("Dataset information:")
print(mobile.info())

print("\nNumerical summary:")
display(mobile.describe())

print("\nRegions:")
print(mobile["Entity"].unique())

print("\nYear range:")
print(mobile["Year"].min(), "to", mobile["Year"].max())

print("\nRecords per region:")
print(mobile["Entity"].value_counts())
```

```python
print("Missing values:")
print(mobile.isnull().sum())

print("\nDuplicate rows:", mobile.duplicated().sum())

print("\nDuplicate region-year combinations:",
      mobile.duplicated(subset=["Entity", "Year"]).sum())

print("\nNegative account values:",
      (mobile["Active mobile money accounts"] < 0).sum())
```

```python
mobile_clean = mobile.copy()

mobile_clean["Year"] = mobile_clean["Year"].astype(int)

mobile_clean = mobile_clean.sort_values(
    by=["Entity", "Year"]
).reset_index(drop=True)

print("Cleaned shape:", mobile_clean.shape)

display(mobile_clean.head(10))
```

```python
mobile_clean.to_csv(
    "../data/processed/mobile_money_cleaned.csv",
    index=False
)

print("Mobile money dataset cleaned and saved successfully!")
```

```python
findex_columns = [
    "countrynewwb",
    "codewb",
    "year",
    "pop_adult",
    "regionwb24_hi",
    "incomegroupwb24",
    "group",
    "group2",
    "account_t_d",
    "fiaccount_t_d",
    "mobileaccount_t_d",
    "dig_acc"
]

findex_selected = findex[findex_columns].copy()

print("Original shape:", findex.shape)
print("Selected shape:", findex_selected.shape)

display(findex_selected.head())
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": findex_selected.isnull().sum(),
    "Missing Percentage": (
        findex_selected.isnull().mean() * 100
    ).round(2)
})

missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_summary)
```

```python
print("Available years:")
print(sorted(findex_selected["year"].dropna().unique()))

print("\nPopulation groups:")
print(findex_selected["group"].value_counts(dropna=False))

print("\nSubgroups:")
print(findex_selected["group2"].value_counts(dropna=False))

print("\nNumber of countries:")
print(findex_selected["countrynewwb"].nunique())

print("\nData types:")
print(findex_selected.dtypes)
```

```python
pakistan = findex_selected[
    findex_selected["countrynewwb"].str.strip().str.lower() == "pakistan"
].copy()

print("Pakistan records:", pakistan.shape[0])

display(
    pakistan.sort_values(
        by=["year", "group", "group2"]
    )
)
```

```python
duplicate_check = findex_selected.duplicated(
    subset=["codewb", "year", "group", "group2"]
)

print("Duplicate country-year-group combinations:",
      duplicate_check.sum())
```

```python

findex_clean = findex_selected.copy()

text_columns = [
    "countrynewwb",
    "codewb",
    "regionwb24_hi",
    "incomegroupwb24",
    "group",
    "group2"
]

for column in text_columns:
    findex_clean[column] = findex_clean[column].str.strip()


numeric_columns = [
    "year",
    "pop_adult",
    "account_t_d",
    "fiaccount_t_d",
    "mobileaccount_t_d",
    "dig_acc"
]

for column in numeric_columns:
    findex_clean[column] = pd.to_numeric(
        findex_clean[column], errors="coerce"
    )


findex_clean["year"] = findex_clean["year"].astype(int)


findex_clean = findex_clean.sort_values(
    by=["countrynewwb", "year", "group", "group2"]
).reset_index(drop=True)


print("Cleaned shape:", findex_clean.shape)
print("Missing values:", findex_clean.isnull().sum().sum())
print("Duplicate rows:", findex_clean.duplicated().sum())

display(findex_clean.head())
```

```python

findex_clean.to_csv(
    "../data/processed/global_findex_cleaned.csv",
    index=False
)

print("Global Findex dataset cleaned and saved successfully!")
```
